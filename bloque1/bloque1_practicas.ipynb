## Preparación

Descarga los materiales del Bloque 1 (el mismo ZIP que en clase) y comprueba que todo está listo.
Es el equivalente a `COMPROBAR_BLOQUE1.bat`.

```python
# PREPARACIÓN · ejecuta esta celda primero
import os, zipfile, urllib.request, subprocess, sys
from pathlib import Path

ZIP_URL = "https://raw.githubusercontent.com/consumer-spain/curso-ia-colab/main/bloque1/BLOQUE1_MATERIALES.zip"
CARPETA = "BLOQUE1_MATERIALES_ACTUALIZADOS"
TRABAJO = Path.cwd().resolve()
if CARPETA in TRABAJO.parts:      # si se vuelve a ejecutar esta celda desde una carpeta del material, no se anida otra copia
    TRABAJO = Path(*TRABAJO.parts[:TRABAJO.parts.index(CARPETA)])
BASE = TRABAJO / CARPETA

if not BASE.exists():
    print("Descargando los materiales del Bloque 1...")
    urllib.request.urlretrieve(ZIP_URL, TRABAJO / "materiales_b1.zip")
    with zipfile.ZipFile(TRABAJO / "materiales_b1.zip") as z:
        z.extractall(TRABAJO)
    print("Materiales listos en:", BASE)
else:
    print("Los materiales ya estaban descargados:", BASE)

# Paquetes (normalmente Colab ya los trae)
for paquete in ("pandas", "openpyxl"):
    try:
        __import__(paquete)
    except ImportError:
        print("Instalando", paquete, "...")
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", paquete], check=True)

def ir_a(carpeta):
    """Se coloca en una carpeta del material (como abrir la carpeta en el Explorador)."""
    os.chdir(BASE / carpeta if carpeta else BASE)
    print("Estás en:", Path.cwd().relative_to(TRABAJO))

ir_a("")
print()
!python 00_comprobacion/comprobador_b1.py
```

# Misión 4 · Concretar

Esta misión es sobre todo una **conversación con la IA**, así que funciona igual en cualquier sitio.

**Una necesidad real:** quieres saber cuál de dos productos sale realmente más barato.
Producto A: 3,60 € · 600 g. Producto B: 5,20 € · 1.000 g. Precios y cantidades distintos.

Y le escribes a la IA: *«Hazme un programa para comparar precios.»*

**¿Qué le falta para poder hacerlo bien?** Piénsalo antes de seguir.

### ¿Qué le falta?

| | Pregunta | Ejemplo en este caso |
|---|---|---|
| ENTRADAS | ¿Qué datos tendrá? | precio y gramos de A y B |
| REGLAS | ¿Cómo los compara? | €/kg |
| RESULTADO | ¿Qué debe devolver? | ambos €/kg + cuál es más barato + diferencia |
| USO | ¿Cómo lo utilizaremos? | en la terminal |
| ERRORES | ¿Qué puede salir mal? | cantidad 0 o negativa |
| COMPROBACIÓN | ¿Cómo sabemos que funciona? | con un caso conocido |

*Son preguntas útiles para este problema, no una plantilla obligatoria.*

### ¿No sabes qué falta? Pregúntale a la IA

> Antes de programar, hazme las preguntas que necesites.

Responde a sus preguntas. Al final, **ya sabemos que el programa debe…**
- pedir nombre, precio y gramos de dos productos
- calcular el precio por kilogramo
- mostrar el €/kg de ambos, cuál es más barato y la diferencia
- avisar y no comparar si una cantidad no es válida

…y la conversación sigue simplemente con: *«Perfecto. Ahora genera el programa con lo que hemos definido.»*

Esta es una versión que una IA puede construir con ese encargo. Ejecútala con el caso de control
